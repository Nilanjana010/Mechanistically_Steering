```python
import pickle
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from collections import Counter
```

```python
with open("category_dict", "rb") as f:
    category_dict = pickle.load(f)
```

```python
category_dict
```

```python
       
df = pd.DataFrame.from_dict(Counter({23: 116, 21: 113, 24: 111, 20: 111, 22: 108, 19: 107, 18: 101, 25: 93, 15: 88, 17: 82, 13: 79, 16: 74, 14: 64, 12: 58, 7: 55, 8: 44, 11: 43, 6: 41, 10: 40, 9: 38, 5: 28, 1: 26, 3: 22, 4: 21, 2: 18, 0: 10}), orient='index', columns=['count'])
df.index.name = 'Gemma-2-2B Model Layers'
df = df.sort_index()

plt.figure(figsize=(14, 1.4))
sns.heatmap(df.T, annot=True, fmt='d', cmap='YlOrRd', cbar=True, linewidths=0.5)
plt.title("Overall Heatmap")
plt.tight_layout()
plt.show()
```

```python
for k_2, v_2 in category_dict.items():
    lst = []
    for data_file_2, layer_val in v_2.items():    
        print(layer_val)
        
        s1 = pd.Series(layer_val, name=data_file_2[8:-4])
        lst.append(s1)
    # value is the layer occurs or is steerable across how many prompts in that category
    df = pd.concat([x.to_frame().T for x in lst], axis=0).fillna(0).astype(int)
    df.index.name = 'Harmful Category'
    df = df.sort_index(axis=1)

    plt.figure(figsize=(12, 5))
    sns.heatmap(df, annot=True, fmt='d', cmap='YlOrRd', cbar=True, linewidths=0.5)
    plt.title(k_2.title() + " Heatmap")
    plt.tight_layout()
    plt.show()
```
